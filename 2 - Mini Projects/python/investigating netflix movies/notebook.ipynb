```python
# Importing pandas, numpy, and matplotlib
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Read in the Netflix CSV as a DataFrame
netflix_df = pd.read_csv("netflix_data.csv")
```

```python
# Checking out the dataset
print(netflix_df[['type', 'release_year', 'duration', 'genre']])
print(netflix_df.columns)
print(netflix_df.dtypes)
print(netflix_df['type'].unique())
print(netflix_df['genre'].unique())
print(netflix_df.shape)
```

```python
# First, I will create a subset of movies that were released between 1990 and 1999, from the netflix_df dataset
movies_only = netflix_df[netflix_df['type'] == 'Movie']
netflix_of_90s = movies_only[np.logical_and(movies_only['release_year'] >= 1990, movies_only['release_year'] <= 1999)]
print(netflix_of_90s.shape)
print(netflix_of_90s[['release_year', 'type', 'genre', 'duration']])
```

```python
# Next, I will find the most frequent movie duration
netflix_of_90s['duration'].hist(bins=15)
plt.show()
duration = 100
```

```python
# Lastly, I will count the number of short actions movies from the 1990s
action_movies_90s = netflix_of_90s[netflix_of_90s['genre'] == 'Action']
short_movie_count = 0
for label, row in action_movies_90s.iterrows():
    if row['duration'] < 90:
        short_movie_count += 1

print(short_movie_count)
```
